##Is Florida getting warmer?##

First, remove all variables in the R environment. Then, load and check the data frame containing temperature from 1901 to 2000 in Florida.

```python
rm(list=ls())
```

```python
load("../data/KeyWestAnnualMeanTemperature.RData")
```

```python
ls()
class(ats)
head(ats)
plot(ats)
```

Observed correlation coefficient between Year and Temp can be calculated but not usable as measurements of climatic variables in successive time-points in a time series (successive seconds, minutes, hours, months, years, etc.) are not independent. Therefore, repeat this calculation a sufficient number of times, each time randomly reshuffling the temperatures , and recalculating the correlation coefficient as shown in the following code:

```python
# Observed correlation between Year and Temp
obs_corr <- cor(ats$Year, ats$Temp)
```

```python
# Permutation test for cor between Year and reshuffled Temp (n=10k)
n_permutations <- 10000
perm_corrs <- matrix()
for (i in 1:n_permutations) {
  perm_temps <- sample(ats$Temp, replace = FALSE) 
  perm_corrs[i] <- cor(ats$Year, perm_temps)
}
```

Now, we can compare the random correlation coefficients with that observed one to get an approximate, asymptotic p-value:

```python
# Two sided p-value as the fraction of permutation larger or equal to observed correlation coefficient 
p_value <- mean(abs(perm_corrs) >= abs(obs_corr))
```

P-value is extremely close to zero which indicates a highly significant warming trend in Florida over the last century.
